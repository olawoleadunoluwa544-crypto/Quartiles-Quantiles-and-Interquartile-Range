```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Titanic Dataset.csv')
df.head()
```

```python
mean_age=np.mean(df['Age'])
print("mean age:", mean_age)
```

```python
median_age=np.median(df['Age'])
print("median age: ", median_age)
```

```python
age_q1=np.quantile(df['Age'],0.25)
age_q2=np.quantile(df['Age'],0.50)
age_q3=np.quantile(df['Age'],0.75)

print(f"Q1: {age_q1}")
print(f"Q2: {age_q2}")
print(f"Q3: {age_q3}")
```

```python
IQR=age_q3-age_q1
print(f"IQR: {IQR}")
```

```python
plt.boxplot(df['Age'])
plt.title('Age Distribution')
plt.show()
```

```python
median_fare=np.median(df['Fare'])
print("Median of Fare:", median_fare)
```

```python
fare_q1=np.quantile(df['Fare'], 0.25)
fare_q2=np.quantile(df['Fare'], 0.50)
fare_q3=np.quantile(df['Fare'], 0.75)

print(f"Q1 {fare_q1}")
print(f"Q2 {fare_q2}")
print(f"Q3 {fare_q3}")
```

```python
IQR=fare_q3-fare_q1
print(f"IQR: {IQR}")
```

```python
plt.boxplot(df['Fare'])
plt.title('Distribution of fare')
plt.show()
```
